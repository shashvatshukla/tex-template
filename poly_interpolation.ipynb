```python
import numpy as np
import matplotlib.pyplot as plt

# --- Define two different cubic polynomials ---
def p(x):
    # p(x) = 0.2(x-1)(x-2)(x-4) + 2
    return 0.2 * (x - 1) * (x - 2) * (x - 4) + 2

def q(x):
    # q(x) = -0.15(x-0.5)(x-3)(x-4) + 1
    return -0.15 * (x - 0.5) * (x - 3) * (x - 4) + 1

# Domain for plotting the full curves
x = np.linspace(-1, 5, 400)

# Four distinct evaluation points
x_samples = np.array([-0.5, 1.0, 2.5, 4.0])
p_samples = p(x_samples)
q_samples = q(x_samples)

# --- Plotting ---
fig, ax = plt.subplots(figsize=(8, 5))

# Plot the two cubic polynomials
ax.plot(x, p(x), label=r"$p(x)$ (cubic 1)", linewidth=2)
ax.plot(x, q(x), label=r"$q(x)$ (cubic 2)", linewidth=2, linestyle="--")

# Vertical dashed lines at the four sample points
for xs in x_samples:
    ax.axvline(xs, linestyle=":", color="gray", alpha=0.7)

# Mark the sampled points on each curve
ax.scatter(x_samples, p_samples, s=50, zorder=3, label=r"$p(x_i)$ samples")
ax.scatter(x_samples, q_samples, s=50, zorder=3, marker="s", label=r"$q(x_i)$ samples")

# Cosmetic stuff
ax.set_xlabel(r"$x$")
ax.set_ylabel(r"$y$")
ax.set_title("Two cubic polynomials evaluated at four distinct points")
ax.legend(loc="best")
ax.grid(alpha=0.2)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# --- Define one quadratic and one cubic polynomial ---
def q(x):
    # Quadratic: q(x) = 0.5(x-1)(x-4) + 1
    return 0.5*(x - 1)*(x - 4) + 1

def c(x):
    # Cubic: c(x) = -0.1(x-0.5)(x-2)(x-4) + 2
    return -0.1*(x - 0.5)*(x - 2)*(x - 4) + 2

# Domain for plotting
x = np.linspace(-1, 5, 400)

# Four sample points
x_samples = np.array([-0.5, 1.0, 2.5, 4.0])
q_samples = q(x_samples)
c_samples = c(x_samples)

# --- Plotting ---
fig, ax = plt.subplots(figsize=(8, 5))

# Plot quadratic (solid)
ax.plot(x, q(x), linewidth=2, label=r"Quadratic $q(x)$")

# Plot cubic (dashed)
ax.plot(x, c(x), linewidth=2, label=r"Cubic $c(x)$")

# Vertical dashed sampling lines
for xs in x_samples:
    ax.axvline(xs, linestyle=":", color="gray", alpha=0.7)

# Mark evaluated points
ax.scatter(x_samples, q_samples, color="C0", s=50, zorder=3, label=r"$q(x_i)$ samples")
ax.scatter(x_samples, c_samples, color="C1", s=50, marker="s", zorder=3, label=r"$c(x_i)$ samples")

# Cosmetics
ax.set_xlabel(r"$x$")
ax.set_ylabel(r"$y$")
ax.set_title("Quadratic vs. Cubic evaluated at four distinct x-points")
ax.legend()
ax.grid(alpha=0.2)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# --- Define the polynomials ---
def q(x):
    return 0.4*(x-1)*(x-3) + 1.5   # quadratic

def c(x):
    return -0.15*(x-0.2)*(x-2)*(x-3.5) + 2   # cubic

# Domain
x = np.linspace(-1, 4, 400)

# Sample points
x_samples = np.array([0, 1, 2, 3])
q_samples = q(x_samples)
c_samples = c(x_samples)

# --- Plotting ---
fig, ax = plt.subplots(figsize=(9, 6))

# Curves
ax.plot(x, q(x), linewidth=2, color="C0")
ax.plot(x, c(x), linewidth=2, linestyle="--", color="C1")

# Label curves directly on the figure
ax.text(3.7, q(3.7), "Quadratic  $q(x)$", color="C0", fontsize=12)
ax.text(3.7, c(3.7), "Cubic  $c(x)$", color="C1", fontsize=12)

# Vertical dashed sampling lines
for xs in x_samples:
    ax.axvline(xs, linestyle=":", color="gray", alpha=0.7)

# Sample point markers
ax.scatter(x_samples, q_samples, color="C0", s=60, zorder=3)
ax.scatter(x_samples, c_samples, color="C1", marker="s", s=60, zorder=3)

# Labels for q(i) and c(i)
for xi, yi in zip(x_samples, q_samples):
    ax.text(xi + 0.05, yi + 0.05, rf"$q({xi})$", color="C0", fontsize=10)

for xi, yi in zip(x_samples, c_samples):
    ax.text(xi + 0.05, yi - 0.12, rf"$c({xi})$", color="C1", fontsize=10)

# Axes labels, grid, title
ax.set_xlabel("x", fontsize=12)
ax.set_ylabel("y", fontsize=12)
ax.set_title("Quadratic and Cubic Evaluated at Four Points (0, 1, 2, 3)", fontsize=14)
ax.grid(alpha=0.2)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# --- Define p1 (cubic) and p2 (quadratic) ---
def p1(x):
    # cubic: p1(x) = x^3 - 3x^2 + 2x + 1
    return x**3 - 3*x**2 + 2*x + 1

def p2(x):
    # quadratic: p2(x) = -x^2 + 4x + 4
    return -x**2 + 4*x + 4

# Domain
x = np.linspace(-1, 4, 400)

# Sample points
x_samples = np.array([0, 1, 2, 3])
p1_samples = p1(x_samples)
p2_samples = p2(x_samples)

# --- Plot ---
fig, ax = plt.subplots(figsize=(10, 6))

# Plot p1 and label
ax.plot(x, p1(x), linewidth=2, color="C0")
ax.text(3.2, p1(3.2)+5,
        r"$p_1(x) = x^3 - 3x^2 + 2x + 1$",
        fontsize=12, color="C0")

# Plot p2 and label
ax.plot(x, p2(x), linewidth=2, linestyle="--", color="C1")
ax.text(3.2, p2(3.2)-5,
        r"$p_2(x) = -x^2 + 4x + 4$",
        fontsize=12, color="C1")

# Vertical dashed sampling lines
for xs in x_samples:
    ax.axvline(xs, linestyle=":", color="gray", alpha=0.7)

# Sample points
ax.scatter(x_samples, p1_samples, color="C0", s=70, zorder=3)
ax.scatter(x_samples, p2_samples, color="C1", marker="s", s=70, zorder=3)

# Label sample points (i, p1(i)) and (i, p2(i))
for xi, yi in zip(x_samples, p1_samples):
    ax.text(xi + 0.05, yi + 0.8,
            rf"$( {xi},\, p_1({xi}) )$",
            color="C0", fontsize=11)

for xi, yi in zip(x_samples, p2_samples):
    ax.text(xi + 0.05, yi - 1.2,
            rf"$( {xi},\, p_2({xi}) )$",
            color="C1", fontsize=11)

# Axes labels and appearance
ax.set_xlabel("x", fontsize=12)
ax.set_ylabel("y", fontsize=12)
ax.set_title("Cubic $p_1$ and Quadratic $p_2$ Evaluated at Points 0, 1, 2, 3", fontsize=14)
ax.grid(alpha=0.2)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# --- Define p1 (cubic) and p2 (quadratic) ---
def p1(x):
    return x**3 - 3*x**2 + 2*x + 1   # cubic

def p2(x):
    return -x**2 + 4*x + 4           # quadratic

# Domain
x = np.linspace(-1, 4, 400)

# Sample points
x_samples = np.array([0, 1, 2, 3])
p1_samples = p1(x_samples)
p2_samples = p2(x_samples)

# --- Plot ---
fig, ax = plt.subplots(figsize=(10, 6))

# Plot p1 and label
ax.plot(x, p1(x), linewidth=2, color="C0")
ax.text(3.1-0.8, p1(3.2)+8,
        r"$p_1(x) = x^3 - 3x^2 + 2x + 1$",
        fontsize=12, color="C0")

# Plot p2 and label
ax.plot(x, p2(x), linewidth=2, color="C1")
ax.text(3.1, p2(3.2)-5,
        r"$p_2(x) = -x^2 + 4x + 4$",
        fontsize=12, color="C1")

# Vertical dashed sampling lines
for xs in x_samples:
    ax.axvline(xs, linestyle=":", color="gray", alpha=0.7)

# Plot all sample points as BLACK X’s
ax.scatter(x_samples, p1_samples, color="black", marker="x", s=80, zorder=3)
ax.scatter(x_samples, p2_samples, color="black", marker="x", s=80, zorder=3)

# --- Label points BELOW (x, y) ---

# Keep track of which labels we've printed (for shared points)
printed = set()

for xi, yi in list(zip(x_samples, p1_samples)) + list(zip(x_samples, p2_samples)):
    
    key = (xi, yi)
    if key in printed:
        continue  # avoid duplicates (e.g., (3,7))
    
    printed.add(key)
    
    ax.text(xi, yi - 1.2,       # BELOW the point
            f"({xi}, {yi})",
            ha='center', va='top',
            color="black", fontsize=11)

# Axis labels and title
ax.set_xlabel("x", fontsize=12)
ax.set_ylabel("y", fontsize=12)
ax.set_title("Illustration of point-value representation", fontsize=14)
ax.grid(alpha=0.2)

plt.tight_layout()
plt.show()
```
